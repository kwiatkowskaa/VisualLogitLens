```python
import os
import shutil
from google.colab import drive
```

```python
def filter_imagenet_custom(original_data_path, filtered_data_path, selected_classes):
    train_subfolders = [f"train.X{i}" for i in range(1, 5)]
    val_folder = "val.X"

    train_filtered_path = os.path.join(filtered_data_path, "train")
    val_filtered_path = os.path.join(filtered_data_path, "val")

    os.makedirs(train_filtered_path, exist_ok=True)
    os.makedirs(val_filtered_path, exist_ok=True)

    for cls in selected_classes:
        in_train = any(os.path.exists(os.path.join(original_data_path, tsf, cls)) for tsf in train_subfolders)
        in_val = os.path.exists(os.path.join(original_data_path, val_folder, cls))

        if in_train and in_val:
            cls_train_dest = os.path.join(train_filtered_path, cls)
            os.makedirs(cls_train_dest, exist_ok=True)

            for train_subfolder in train_subfolders:
                orig_cls_path = os.path.join(original_data_path, train_subfolder, cls)
                if os.path.exists(orig_cls_path):
                    for filename in os.listdir(orig_cls_path):
                        src_file = os.path.join(orig_cls_path, filename)
                        dst_file = os.path.join(cls_train_dest, filename)
                        if not os.path.exists(dst_file):
                            shutil.copy2(src_file, dst_file)
                    print(f"Copied train class '{cls}' from {train_subfolder}")

            orig_cls_val_path = os.path.join(original_data_path, val_folder, cls)
            cls_val_dest = os.path.join(val_filtered_path, cls)
            shutil.copytree(orig_cls_val_path, cls_val_dest, dirs_exist_ok=True)
            print(f"Copied val class '{cls}'")

        else:
            print(f"Skipping class '{cls}' - train folder present: {in_train}, val folder present: {in_val}")
```
